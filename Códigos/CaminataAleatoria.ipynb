# Caminata Aleatoria

Una caminata aleatoria es la simulación del recorrido que hace un objeto o una persona cuando cada nuevo paso se toma al azar, sin recordar los pasos anteriores.

**Preparación del entorno**
Importamos las herramientas que nos permiten generar números al azar, operar con listas numéricas y dibujar las trayectorias:

```python
import random as rd
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

Estabelecemos cuántos pasos durará la simulación:

```python
N=10000
```

**Caminata en 1D**: Aquí unicamente puede vanzar o retroceder una unidad.

Es como lanzar una moneda para dar un paso a la derecha (+1) o a la izquierda (-1) sobre una linea recta.


```python
posiciones=[0]

for i in range (1, N+1):
  paso = rd.choice([-1,1])
  nueva_posicion = posiciones[-1] + paso
  posiciones.append(nueva_posicion)
```

```python
plt.figure(figsize=(10, 6))
plt.plot(posiciones)
plt.title("Caminata Aleatoria en 1D")
plt.xlabel("Paso / Tiempo")
plt.ylabel("Posición")
plt.show()
```

**Caminata en 2D**: Se puede decidir si nos movemos a la izquierda/abajo (-1), quedarnos en el lugar (0) o vamos a la derecha/arriba (+1).

Es como una hormiga moviendose sobre una mesa cuadriculada, pudiendo avanzar, retroceder o mantenerse en cada coordenada.

```python
pasos_x = [rd.choice([-1, 0, 1]) for _ in range(N)]
pasos_y = [rd.choice([-1, 0, 1]) for _ in range(N)]

x = np.cumsum(pasos_x)
y = np.cumsum(pasos_y)

plt.figure(figsize=(10, 6))
plt.plot(x, y)
```
