# Semana 2: Filtraciones y la Propiedad de Markov con Arrays Multidimensionales

```python
import numpy as np
import matplotlib.pyplot as plt
```

Definimos parámetros

```python
# Definimos el escenario temporal (filas de la matriz)
T = 1.0
N = 500
dt = T / N

# Definimos el escenario de información (columnas de la matriz)
M = 100 # Vamos a simular 100 realidades paralelas em el tiempo
```

Generamos los arrays multidimensionales simulando flujos de información

```python
tiempo = np.linspace(0, T, N+1) # Creamos el vector de tiempo

incrementos_matriz = np.random.normal(0, np.sqrt(dt), size=(N,M)) #Generamos la matriz de incrementos aleatorios de tamaño (N,M)

fila_inicial = np.zeros((1, M)) #Creamos una fila de ceros para el tiempo inicial en las M trayectorias

matriz_cero = np.vstack([fila_inicial, incrementos_matriz]) # Unimos la fila de ceros con la matriz de incremento

W_multidimensional = np.cumsum(matriz_cero, axis = 0) # Aplicamos la suma acumulada a lo largo del eje del tiempo, esto construye el historial de informacion contiguo
```

```python
# Creamos la ventana de la gráfica
plt.figure(figsize=(11, 5))

# Graficamos el tiempo contra las 100 trayectorias simultáneas
plt.plot(tiempo, W_multidimensional, linewidth=0.8, alpha=0.4) # Usamos un alpha bajo (0.4) para que las líneas sean semitransparentes y podamos ver la densidad

# Detalles de la gráfica
plt.title("Simulación de un Flujo de Información Contiguo: $M=100$ Trayectorias (Semana 2)", fontsize=12, fontweight='bold')
plt.xlabel("Tiempo ($t$)", fontsize=10)
plt.ylabel("Estado del Proceso ($W_t$)", fontsize=10)

# Líneas de referencia para denotar el origen de la Filtración F_0
plt.axhline(0, color='black', linewidth=1.2, linestyle='-')
plt.axvline(0, color='black', linewidth=1.2, linestyle='-')

plt.grid(True, linestyle=':', alpha=0.6) # Cruadícula

plt.show() # Mostrar la gráfica
```
