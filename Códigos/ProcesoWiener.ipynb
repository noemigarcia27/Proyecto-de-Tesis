# Semana 1: Simulación de un Proceso Estocastico

```python
import numpy as np
import matplotlib.pyplot as plt
```

Definimos variables del espacio temporal

```python
T = 1.0     #Tiempo final
N = 1000    #Número de pasos

dt = T / N  #Tamaño del paso

sigma_dt = np.sqrt(dt) #la desviación estandar de los incrementos
```

Construimos el espacio

```python
tiempo = np.linspace(0, T, N+1)  #Eje X
incrementos = np.random.normal(0, sigma_dt, N) #Generar los N incrementos normales

incrementos_cero = np.insert(incrementos, 0, 0) #Iniciamos en 0 para que siga la primera propiedad del proceso

W = np.cumsum(incrementos_cero)
```

Visualización del Proceso de Wiener

```python
plt.Figure(figsize=(10,4.5)) #Crear la ventana de la grafica
plt.plot(tiempo, W, color="blue", linewidth=1.2, label="Trayectoria $W_t$") #Graficar el eje de tiempo contra el eje de las posiciones

#Detalles de la gráfica
plt.title("Simulación de una Trayectoria del Proceso de Wiener Estandar", fontsize=12, fontweight="bold")
plt.xlabel("Tiempo ($t$)", fontsize=10)
plt.ylabel("Posición ($W_t$)", fontsize=10)
plt.axhline(0, color='black', linewidth=0.8, linestyle='--') # Línea de referencia en cero
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()

plt.show() #Mostrar el resultado
```
